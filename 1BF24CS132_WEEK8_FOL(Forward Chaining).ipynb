```python
# Forward Chaining Algorithm

# Initial Knowledge Base (KB)
KB = {
    ("American", "Robert"),
    ("Missile", "T1"),
    ("Sells", "Robert", "T1", "A1"),
    ("Hostile", "A1")
}

# Goal
goal = ("Criminal", "Robert")

print("FORWARD CHAINING ALGORITHM")

# Forward chaining process
while True:
    new_facts = []

    # Rule 1: Missile(x) -> Weapon(x)
    for fact in KB:
        if fact[0] == "Missile":
            x = fact[1]
            derived = ("Weapon", x)

            if derived not in KB:
                new_facts.append(derived)

    # Rule 2:
    # American(x) AND Weapon(y) AND
    # Sells(x,y,z) AND Hostile(z) -> Criminal(x)

    for american in KB:
        if american[0] == "American":
            x = american[1]

            for weapon in KB:
                if weapon[0] == "Weapon":
                    y = weapon[1]

                    for sells in KB:
                        if sells[0] == "Sells":
                            seller, item, target = sells[1:]

                            if seller == x and item == y:
                                hostile = ("Hostile", target)

                                if hostile in KB:
                                    derived = ("Criminal", x)

                                    if derived not in KB:
                                        new_facts.append(derived)

    # If no new facts are derived, stop
    if not new_facts:
        break

    # Add new facts to KB
    for fact in new_facts:
        KB.add(fact)
        print("Derived New Fact:", fact)

# Check the goal
if goal in KB:
    print("Goal (Criminal, Robert) is True")
else:
    print("Goal (Criminal, Robert) is False")
```
