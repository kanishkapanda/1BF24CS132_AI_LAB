```python
import ast
import math


def get_leaves(node):
    """Extracts all leaf numbers inside a pruned branch."""
    if isinstance(node, (int, float)):
        return [node]
    leaves = []
    for child in node:
        leaves.extend(get_leaves(child))
    return leaves


def alpha_beta(node, is_max=True, alpha=-math.inf, beta=math.inf, path="Root"):
    # Base case: reached a leaf number
    if isinstance(node, (int, float)):
        return node, f"{path} ({node})"

    best_path = path

    if is_max:
        v = -math.inf
        for i, child in enumerate(node):
            child_path = f"{path} -> Branch {i+1}"
            val, current_path = alpha_beta(
                child, False, alpha, beta, child_path
            )
            if val > v:
                v = val
                best_path = current_path

            # Beta Cutoff
            if v >= beta:
                for j in range(i + 1, len(node)):
                    pruned_path = f"{path} -> Branch {j+1}"
                    print(
                        f"[PRUNED - Beta Cutoff] Path: {pruned_path} | Skipped Leaves: {get_leaves(node[j])}"
                    )
                return v, best_path

            alpha = max(alpha, v)
        return v, best_path

    else:
        v = math.inf
        for i, child in enumerate(node):
            child_path = f"{path} -> Branch {i+1}"
            val, current_path = alpha_beta(child, True, alpha, beta, child_path)
            if val < v:
                v = val
                best_path = current_path

            # Alpha Cutoff
            if v <= alpha:
                for j in range(i + 1, len(node)):
                    pruned_path = f"{path} -> Branch {j+1}"
                    print(
                        f"[PRUNED - Alpha Cutoff] Path: {pruned_path} | Skipped Leaves: {get_leaves(node[j])}"
                    )
                return v, best_path

            beta = min(beta, v)
        return v, best_path


# --- User Input ---
if __name__ == "__main__":
    print(
        "Enter the game tree as nested brackets (e.g., [[5, 6], [7, 4, 5]]):"
    )
    tree_input = input("> ")

    # Safely convert string input into a nested Python list
    tree = ast.literal_eval(tree_input)

    print("\n--- Pruning Trace ---")
    optimal_val, optimal_path = alpha_beta(tree, is_max=True)

    print("\n--- Final Result ---")
    print(f"Optimal Value at Root (MAX): {optimal_val}")
    print(f"Optimal Path: {optimal_path}")
```
