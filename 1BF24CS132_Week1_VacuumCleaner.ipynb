```python
print("VACUUM CLEANER AGENT")

locations = ["A", "B"]
actions = ["Left", "Right", "Pick the Dust", "No-op"]

print("Locations:", " and ".join(locations))
print("Actions:", ", ".join(actions))

while True:
    location = input("\nEnter the location (A/B): ").upper()
    status = input("Enter status (Clean/Dirty): ").capitalize()

    print("Location:", location)
    print("Status:", status)

    if status == "Dirty":
        action = "Pick the Dust"

    elif status == "Clean":
        if location == "A":
            action = "Right"
        elif location == "B":
            action = "Left"
        else:
            action = "No-op"

    else:
        action = "Invalid status"

    print("Action:", action)

    choice = input("\nDo you want to continue (yes/no): ").lower()

    if choice != "yes":
        break
```
